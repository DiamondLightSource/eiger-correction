# Fast EIGER Correction Extraction: Two-Stage Position Scaling

Repeated-angle frames are summed, then detector response and source-position scales are estimated in separate stages. Stage 1 extracts pixel/sensor gains over `FIT_Q_RANGE_NM`, retaining nuisance scales when enabled. Stage 2 freezes that response and refines one source-only scalar per motor position jointly from full-range overlaps; it replaces stage-1 nuisance scales. No profile smoothing, tapering or Q-dependent scale is introduced.

This notebook uses the functions in the `eiger_correction` package. Change settings in the **parameters** cell, or pass them with papermill, e.g.

```bash
papermill notebooks/extract_eiger_corrections.ipynb run.ipynb \
    -p SCAN_PATH /dls/i15-1/data/2026/cm44163-4/i15-1-100605.nxs \
    -p OUTPUT_DIRECTORY /some/output/directory
```

> **Azimuth selection:** `AZIMUTH_RANGE_DEG = (67.5, 112.5)` retains a **single 45° wedge about pyFAI χ = +90°**, the vertical high-angle scattering direction verified from this geometry. The opposite wedge is not included. `None` restores all azimuths. Bounds are inclusive in degrees; `(170, -170)` crosses the ±180° seam. The argument is passed as `PhysicalModel(..., azimuth_range=AZIMUTH_RANGE_DEG)`. Pixel-centre χ is recomputed at **each motor position**, never turned into a detector-fixed mask. Fitting, validation, coverage diagnostics, final integration and the raw-data audit use the same wedge. The geometry preview shows retained pixels at the lowest/highest positions and retained fraction across the scan. Saved fit settings preserve the range for calibration reload; masked output names include the χ bounds to preserve older results. **Rerun the notebook from the start after changing the range.** This reduces azimuth mixing but does not model the anisotropic beam or guarantee its effects vanish; high-angle frames still sample a narrower sector. Narrower wedges reduce statistical and overlap support; existing fit/graph support checks remain unchanged.

> **Grouping:** angles within 0.001° of each group's lowest angle are represented by their arithmetic-mean motor reading, one equally weighted reading per raw frame, never I0-weighted. Set the tolerance to 0 for exact-only grouping. Raw counts and I0 are summed; I0 normalizes intensity, not geometry. A pixel is excluded from a group if any constituent value is nonfinite, negative or saturated. Stage-1 validation splits whole angle groups. Final source-position scales are applied only to this source scan, not transferred to other scans.

Summation preserves raw-count statistics and exposure normalization. Source files are read-only. `USE_ANGLE_AGGREGATION=False` selects the original per-frame path. Full-range detector-response use outside the extraction Q range remains extrapolation, and fitted-scale uncertainty is not included in the propagated Poisson sigma.

## 1–3. Load Inputs, Parse Angle Metadata, and Detect Repeats

Load the NeXus metadata and companion detector/monitor/metadata files through the shared workflow module. The small angle index below avoids loading detector frames into memory just to inspect repetitions.

The figure compares exact-angle repeats with tolerance-group totals, shows every exact angle's offset from its group's minimum against the allowed tolerance, and magnifies the widest-span group. Shaded intervals are **grouping acceptance windows, not measurement uncertainties**: membership requires `angle − group minimum ≤ ANGLE_TOLERANCE_DEG`, not ± tolerance around the mean. The representative angle is the arithmetic mean of all motor readings in the group, independent of I0. If aggregation is disabled, the figure is labelled as a preview only.

````python
raw_scan = workflow.load_scan(SCAN_PATH, DETECTOR_PATH, MONITOR_PATH, METADATA_PATH)
unique_raw_angles, raw_inverse, raw_repeats = np.unique(
    raw_scan.angles, return_inverse=True, return_counts=True
)
repeat_summary = {
    "raw frames": len(raw_scan.angles),
    "unique exact angles": len(unique_raw_angles),
    "duplicate frames beyond first": int(len(raw_scan.angles) - len(unique_raw_angles)),
    "maximum repeats at one angle": int(raw_repeats.max()),
    "angles repeated more than once": int(np.count_nonzero(raw_repeats > 1)),
}
print("Raw scan summary:", repeat_summary)
print(
    "Angle range (degrees):", float(raw_scan.angles.min()), float(raw_scan.angles.max())
)
repeat_distribution = np.unique(raw_repeats, return_counts=True)
print(
    "Repeat-count distribution (frames per angle -> number of angles):",
    {
        int(count): int(number)
        for count, number in zip(*repeat_distribution, strict=True)
    },
)
print("First source frame -> exact-angle index/repeat count:")
for frame_index in range(min(12, len(raw_scan.angles))):
    group_index = int(raw_inverse[frame_index])
    print(
        frame_index,
        float(raw_scan.angles[frame_index]),
        group_index,
        int(raw_repeats[group_index]),
    )

# Metadata-only preview: use exactly the minimum-anchored rule in
# workflow.aggregate_equal_angle_frames, without reading detector images.
if not np.isfinite(ANGLE_TOLERANCE_DEG) or ANGLE_TOLERANCE_DEG < 0:
    raise ValueError("ANGLE_TOLERANCE_DEG must be finite and nonnegative")
preview_clusters = []
for angle_index, angle in enumerate(unique_raw_angles):
    if (
        not preview_clusters
        or angle - unique_raw_angles[preview_clusters[-1][0]] > ANGLE_TOLERANCE_DEG
    ):
        preview_clusters.append([angle_index])
    else:
        preview_clusters[-1].append(angle_index)
preview_group_for_angle = np.empty(len(unique_raw_angles), dtype=int)
for group_index, cluster in enumerate(preview_clusters):
    preview_group_for_angle[cluster] = group_index
preview_group_for_frame = preview_group_for_angle[raw_inverse]
preview_counts = np.bincount(preview_group_for_frame)
# Each motor reading contributes equally; I0 does not enter the angle mean.
preview_means = np.array(
    [
        np.mean(raw_scan.angles[preview_group_for_frame == group_index])
        for group_index in range(len(preview_clusters))
    ]
)
preview_minima = np.array(
    [unique_raw_angles[cluster[0]] for cluster in preview_clusters]
)
preview_spreads = np.array(
    [np.ptp(unique_raw_angles[cluster]) for cluster in preview_clusters]
)
preview_offsets_mdeg = 1000 * (
    unique_raw_angles - preview_minima[preview_group_for_angle]
)
tolerance_mdeg = 1000 * ANGLE_TOLERANCE_DEG
preview_numbers = np.arange(1, len(preview_clusters) + 1)
preview_status = (
    "Aggregation enabled"
    if USE_ANGLE_AGGREGATION
    else "PREVIEW ONLY — aggregation disabled"
)
print(
    f"Tolerance-group preview: {len(unique_raw_angles)} exact angles -> "
    f"{len(preview_clusters)} groups; "
    f"maximum span = {preview_spreads.max():.6g} degrees"
)

fig = plt.figure(figsize=(15, 9), constrained_layout=True)
grid = fig.add_gridspec(2, 2, height_ratios=(1, 1.2), width_ratios=(1.6, 1))
ax_counts = fig.add_subplot(grid[0, :])
ax_offsets = fig.add_subplot(grid[1, 0])
ax_zoom = fig.add_subplot(grid[1, 1])
fig.suptitle(
    f"Angle repeats and grouping tolerance: {ANGLE_TOLERANCE_DEG:g}° "
    f"({tolerance_mdeg:g} mdeg)\n"
    f"{preview_status} | {len(raw_scan.angles):,} frames → "
    f"{len(preview_clusters)} tolerance groups",
    fontsize=14,
)

ax_counts.vlines(unique_raw_angles, 0, raw_repeats, color="0.65", alpha=0.6)
ax_counts.scatter(
    unique_raw_angles,
    raw_repeats,
    s=22,
    color="0.35",
    label="Frames at each exact angle",
    zorder=3,
)
ax_counts.plot(
    preview_means,
    preview_counts,
    "o",
    color="tab:blue",
    markersize=5,
    label="Total frames per tolerance group (at mean motor angle)",
)
ax_counts.set(
    title="A · Exact repeats versus tolerance-group totals",
    xlabel="2θ (degrees)",
    ylabel="Raw frames",
)
ax_counts.set_ylim(bottom=0)
ax_counts.legend(loc="best")

# Millidegree offsets make the small acceptance windows visible over a wide scan.
if tolerance_mdeg > 0:
    ax_offsets.axhspan(
        0,
        tolerance_mdeg,
        color="tab:blue",
        alpha=0.09,
        label="Allowed window from group minimum",
    )
    ax_zoom.axvspan(
        0, tolerance_mdeg, color="tab:blue", alpha=0.09, label="Allowed grouping window"
    )
ax_offsets.axhline(
    tolerance_mdeg,
    color="tab:red",
    linestyle="--",
    label=f"Tolerance = {tolerance_mdeg:g} mdeg",
)
ax_offsets.vlines(
    preview_numbers, 0, 1000 * preview_spreads, color="0.5", linewidth=1.5
)
point_sizes = 18 + 65 * raw_repeats / raw_repeats.max()
ax_offsets.scatter(
    preview_group_for_angle + 1,
    preview_offsets_mdeg,
    s=point_sizes,
    color="0.3",
    alpha=0.7,
    label="Exact angles (size indicates repeat count)",
    zorder=3,
)
ax_offsets.scatter(
    preview_numbers,
    1000 * (preview_means - preview_minima),
    marker="D",
    s=22,
    color="tab:orange",
    label="Mean motor angle (unweighted)",
    zorder=4,
)
ax_offsets.set(
    title="B · Within-group angle spread for every group",
    xlabel="Tolerance group (in increasing angle order)",
    ylabel="Angle − group minimum (mdeg)",
)
ax_offsets.set_xticks(preview_numbers[:: max(1, len(preview_numbers) // 12)])
offset_scale = max(tolerance_mdeg, float(preview_offsets_mdeg.max()), 0.001)
ax_offsets.set_ylim(-0.08 * offset_scale, 1.35 * offset_scale)
ax_offsets.legend(loc="upper left", fontsize=8)

# Deterministically inspect the widest-span group; still works for exact-only grouping.
zoom_group = int(np.argmax(preview_spreads))
zoom_indices = np.asarray(preview_clusters[zoom_group])
zoom_offsets = preview_offsets_mdeg[zoom_indices]
ax_zoom.vlines(zoom_offsets, 0, raw_repeats[zoom_indices], color="0.4", linewidth=2)
ax_zoom.scatter(
    zoom_offsets,
    raw_repeats[zoom_indices],
    color="0.25",
    s=40,
    zorder=3,
    label="Exact-angle repeat counts",
)
ax_zoom.axvline(
    tolerance_mdeg, color="tab:red", linestyle="--", label="Tolerance boundary"
)
ax_zoom.axvline(
    1000 * (preview_means[zoom_group] - preview_minima[zoom_group]),
    color="tab:orange",
    linestyle=":",
    linewidth=2,
    label="Mean motor angle (unweighted)",
)
for offset, count in zip(zoom_offsets, raw_repeats[zoom_indices], strict=True):
    ax_zoom.annotate(
        str(int(count)),
        (offset, count),
        xytext=(4, 6),
        textcoords="offset points",
        fontsize=8,
    )
ax_zoom.set(
    title=(
        f"C · Widest-span group {zoom_group + 1}: {preview_counts[zoom_group]} frames\n"
        f"Minimum 2θ = {preview_minima[zoom_group]:.6f}°; "
        f"span = {1000 * preview_spreads[zoom_group]:.4g} mdeg"
    ),
    xlabel="Angle − group minimum (mdeg)",
    ylabel="Raw frames at exact angle",
)
ax_zoom.set_xlim(-0.08 * offset_scale, 1.15 * offset_scale)
ax_zoom.set_ylim(0, 1.4 * raw_repeats[zoom_indices].max())
ax_zoom.legend(loc="upper right", fontsize=8)
for axis in (ax_counts, ax_offsets, ax_zoom):
    axis.grid(axis="y", alpha=0.2)
    axis.spines[["top", "right"]].set_visible(False)
plt.show()
````

## 4–6. Combine Matching Frames and Rebuild the Reduced Processing Dataset

The reusable helper streams frames into a temporary HDF5 dataset, summing counts and I0 per angle group. For every pixel, any invalid member causes that group's pixel to be masked, rather than silently treating missing data as zero. Downstream fitting continues to use the same `Scan`, cache, fitting, and validation interfaces.

````python
aggregation_seconds = 0.0
if USE_ANGLE_AGGREGATION:
    working_directory = tempfile.TemporaryDirectory(
        dir=OUTPUT_DIRECTORY, prefix=".eiger_fast_"
    )
    AGGREGATED_DATA_PATH = Path(working_directory.name) / "angle_aggregated_frames.h5"
    aggregation_start = time.perf_counter()
    analysis_scan, frame_groups = workflow.aggregate_equal_angle_frames(
        raw_scan,
        AGGREGATED_DATA_PATH,
        saturation=RAW_SATURATION,
        angle_tolerance=ANGLE_TOLERANCE_DEG,
    )
    aggregation_seconds = time.perf_counter() - aggregation_start
else:
    analysis_scan = raw_scan
    frame_groups = [
        np.array([index], dtype=int) for index in range(len(raw_scan.angles))
    ]
    working_directory = None
    AGGREGATED_DATA_PATH = None

print("Reduced frames:", len(analysis_scan.angles))
print("Frame reduction factor:", len(raw_scan.angles) / len(analysis_scan.angles))
group_angle_spreads = np.array(
    [np.ptp(raw_scan.angles[members]) for members in frame_groups]
)
print(
    "Groups combining distinct exact angle values:",
    int(np.count_nonzero(group_angle_spreads > 0)),
)
print(
    "Maximum source-angle span within a group (degrees):",
    float(group_angle_spreads.max()),
)
print("Aggregation elapsed (s):", aggregation_seconds)
print(
    "Reduced detector dataset shape:",
    (len(analysis_scan.angles), *analysis_scan.shape),
)
print("I0 aggregation check:", bool(np.all(analysis_scan.i0 > 0)))
if AGGREGATED_DATA_PATH is not None:
    with h5py.File(AGGREGATED_DATA_PATH, "r") as grouped_file:
        np.testing.assert_equal(
            workflow.h5_dataset(grouped_file, "data").shape,
            (len(frame_groups), *raw_scan.shape),
        )
        np.testing.assert_equal(
            workflow.h5_dataset(grouped_file, "source_frame_indices").shape[0],
            len(raw_scan.angles),
        )
        print("Temporary grouped HDF5:", AGGREGATED_DATA_PATH)

# Check one of the smallest repeated groups without stacking all its raw images.
check_group_index = min(
    (i for i, members in enumerate(frame_groups) if len(members) > 1),
    key=lambda i: len(frame_groups[i]),
    default=None,
)
if check_group_index is not None and AGGREGATED_DATA_PATH is not None:
    members = frame_groups[check_group_index]
    expected = np.zeros(raw_scan.shape, dtype=float)
    invalid_any = np.zeros(raw_scan.shape, dtype=bool)
    with (
        h5py.File(raw_scan.detector_path, "r") as raw_file,
        h5py.File(AGGREGATED_DATA_PATH, "r") as grouped_file,
    ):
        raw_frames = workflow.h5_dataset(raw_file, raw_scan.detector_dataset)
        for frame_index in members:
            raw_frame = np.asarray(raw_frames[int(frame_index)], dtype=float)
            invalid = (
                ~np.isfinite(raw_frame)
                | (raw_frame < 0)
                | (raw_frame >= RAW_SATURATION)
            )
            invalid_any |= invalid
            expected += np.where(invalid, 0.0, raw_frame)
        expected[invalid_any] = np.nan
        np.testing.assert_allclose(
            workflow.h5_dataset(grouped_file, "data")[check_group_index],
            expected,
            rtol=0,
            atol=0,
            equal_nan=True,
        )
        np.testing.assert_allclose(
            analysis_scan.i0[check_group_index],
            raw_scan.i0[members].sum(),
            rtol=1e-14,
        )
    angle_spread = float(np.ptp(raw_scan.angles[members]))
    print(
        "Sample aggregate raw-count and summed-I0 checks passed for",
        len(members),
        "frames; angle spread (degrees):",
        angle_spread,
    )
else:
    print(
        "No repeated group available for the sample check; "
        "dataset shape and I0 checks passed."
    )
````

````python
# Compare separate and summed integrations for a tiny representative sample.
# Prefer a tolerance-merged group that actually contributes in the fitting q range.
tolerance_group_index = None
for i, group in enumerate(frame_groups):
    if np.ptp(raw_scan.angles[group]) <= 0:
        continue
    representative_angle = float(np.mean(raw_scan.angles[group]))
    q_pixels = physical_model.arrays(representative_angle)[1]
    sample_geometry_mask = base_mask | workflow.azimuth_mask(
        physical_model, representative_angle
    )
    q_support = (
        (~sample_geometry_mask)
        & (q_pixels >= FIT_Q_RANGE_NM[0])
        & (q_pixels < FIT_Q_RANGE_NM[1])
    )
    if np.count_nonzero(q_support) >= 100:
        tolerance_group_index = i
        break

if USE_ANGLE_AGGREGATION and tolerance_group_index is not None:
    tolerance_members = frame_groups[tolerance_group_index]
    _, first_positions = np.unique(
        raw_scan.angles[tolerance_members], return_index=True
    )
    sample_indices = tolerance_members[np.sort(first_positions)]
elif (
    USE_ANGLE_AGGREGATION
    and check_group_index is not None
    and len(frame_groups[check_group_index]) <= 4
):
    sample_indices = frame_groups[check_group_index]
else:
    sample_indices = np.array([], dtype=int)

if len(sample_indices) > 1:
    with h5py.File(raw_scan.detector_path, "r") as raw_file:
        raw_frames = workflow.h5_dataset(raw_file, raw_scan.detector_dataset)
        raw_sample_frames = [
            np.asarray(raw_frames[int(frame_index)], dtype=float)
            for frame_index in sample_indices
        ]
    raw_sample_stack = np.stack(raw_sample_frames)
    invalid_any = np.any(
        ~np.isfinite(raw_sample_stack)
        | (raw_sample_stack < 0)
        | (raw_sample_stack >= RAW_SATURATION),
        axis=0,
    )
    common_mask = base_mask | invalid_any
    sample_counts = np.where(invalid_any, 0.0, raw_sample_stack).sum(axis=0)
    sample_counts[invalid_any] = np.nan
    sample_i0 = float(raw_scan.i0[sample_indices].sum())
    sample_angle = float(np.mean(raw_scan.angles[sample_indices]))
    per_frame_results = {
        int(frame_index): workflow.integrate_frame(
            physical_model,
            raw_sample_frames[position],
            raw_scan.angles[int(frame_index)],
            raw_scan.i0[int(frame_index)],
            common_mask,
            np.ones(analysis_scan.shape),
            FIT_BIN_COUNT,
            FIT_Q_RANGE_NM,
            saturation=RAW_SATURATION,
        )
        for position, frame_index in enumerate(sample_indices)
    }
    summed_result = workflow.integrate_frame(
        physical_model,
        sample_counts,
        sample_angle,
        sample_i0,
        common_mask,
        np.ones(analysis_scan.shape),
        FIT_BIN_COUNT,
        FIT_Q_RANGE_NM,
        saturation=np.inf,
    )
    separate_merged = workflow.merge_results(per_frame_results)
    separate_intensity = separate_merged["intensity"]
    valid_bins = (np.asarray(summed_result.count) > 0) & (
        separate_merged["accumulators"][3] > 0
    )
    valid_bins &= np.isfinite(summed_result.intensity) & np.isfinite(separate_intensity)
    valid_bins &= (summed_result.intensity > 0) & (separate_intensity > 0)
    relative_difference = (
        np.abs(summed_result.intensity[valid_bins] - separate_intensity[valid_bins])
        / separate_intensity[valid_bins]
    )
    print(
        "Sample separate-vs-summed integration; frames:",
        len(sample_indices),
        "angle span (deg):",
        float(np.ptp(raw_scan.angles[sample_indices])),
    )
    if relative_difference.size:
        print(
            "Relative intensity difference, median / 95th percentile / max (%):",
            100 * np.percentile(relative_difference, [50, 95, 100]),
        )
    else:
        print("No common supported q bins in the sample integration comparison.")
else:
    print(
        "Integration comparison skipped: no tolerance-merged group contributes "
        "in the fitting q range."
    )
````

## Stage 1 — extract detector response over the selected Q range

`FIT_Q_RANGE_NM` restricts pixel/sensor extraction. `FIT_FRAME_GAINS=True` retains nuisance position scales **during this fit** to reduce leakage into the detector map; only positions represented in the fitting cache have these scales refined. These stage-1 scales are not the final exported scales when `REFINE_ALL_POSITION_SCALES=True`.

The held-out angle-group validation below tests stage 1 only. Its lower-Q-inferred gains are diagnostic, not stage-2 gains. Monitor overlays are independently mean-normalized for display only; measured summed I0 remains the actual exposure normalization. With `FIT_FRAME_GAINS=False`, the stage-1 fit has unity nuisance scales, independently of the stage-2 switch.

````python
cache_start = time.perf_counter()
cache = workflow.build_cache(
    analysis_scan,
    physical_model,
    base_mask,
    FIT_Q_RANGE_NM,
    FIT_BIN_COUNT,
    saturation=cache_saturation,
)
cache_seconds = time.perf_counter() - cache_start
contributing_indices = np.array(sorted(cache), dtype=int)
print("Cache construction time (s):", cache_seconds)
print("Cached contributing angle groups:", len(contributing_indices))
print("Cached observations:", sum(len(frame["pixels"]) for frame in cache.values()))
cache_bytes = sum(array.nbytes for frame in cache.values() for array in frame.values())
print("Sparse cache memory (MiB):", cache_bytes / 2**20)
assert fit_options["fit_frame_gains"] == FIT_FRAME_GAINS, (
    "Rerun the fit-settings cell after changing FIT_FRAME_GAINS."
)

# Half-data validation needs a separate support threshold; do not weaken the full fit.
validation_fit_options = {**fit_options, "minimum_observations": 4}
validation_reports = []
validation_fits = []
if RUN_VALIDATION:
    print(
        "Validation / full-fit minimum observations:",
        validation_fit_options["minimum_observations"],
        fit_options["minimum_observations"],
    )
    even = contributing_indices[contributing_indices % 2 == 0]
    odd = contributing_indices[contributing_indices % 2 == 1]
    folds = (
        ("Even angle groups train / odd test", even, odd),
        ("Odd angle groups train / even test", odd, even),
    )
    for fold_name, train_indices, test_indices in folds:
        print("\n", fold_name)
        fold_fit = workflow.fit_response(
            cache,
            train_indices,
            sensor_ids,
            len(analysis_scan.angles),
            **validation_fit_options,
        )
        if not FIT_FRAME_GAINS:
            np.testing.assert_array_equal(
                fold_fit["frame_gains"], np.ones(len(analysis_scan.angles))
            )
        validation = workflow.validate_response(
            analysis_scan,
            physical_model,
            cache,
            fold_fit,
            test_indices,
            base_mask,
            FIT_Q_RANGE_NM,
            FIT_BIN_COUNT,
            saturation=cache_saturation,
            validate_frame_gains=FIT_FRAME_GAINS,
        )
        validation_reports.append(
            {
                "fold": fold_name,
                "minimum_observations": validation_fit_options["minimum_observations"],
                "validate_frame_gains": bool(FIT_FRAME_GAINS),
                "metrics": validation["summary"],
                "frame_gain_checks": validation["frame_rows"],
            }
        )
        validation_fits.append(fold_fit)
        for row in validation["summary"]:
            print(row)
    fold_sensor_difference = 100 * np.sqrt(
        np.mean(
            (
                np.log(validation_fits[0]["sensor_gains"])
                - np.log(validation_fits[1]["sensor_gains"])
            )
            ** 2
        )
    )
    print("Sensor log-gain RMS difference between folds (%):", fold_sensor_difference)
else:
    print("WARNING: independent angle-group validation is disabled.")

print("\nFull grouped-scan fit; refined position gains:", FIT_FRAME_GAINS)
calibration_fit = workflow.fit_response(
    cache, contributing_indices, sensor_ids, len(analysis_scan.angles), **fit_options
)
if not FIT_FRAME_GAINS:
    np.testing.assert_array_equal(
        calibration_fit["frame_gains"], np.ones(len(analysis_scan.angles))
    )
print("Sensor gains:", calibration_fit["sensor_gains"])
print("Persistent bad-pixel candidates:", int(calibration_fit["bad_pixel_mask"].sum()))
print(
    "Position gain range:",
    float(calibration_fit["frame_gains"].min()),
    float(calibration_fit["frame_gains"].max()),
)

# Ring current is an instantaneous monitor: average within the same raw-frame groups.
with h5py.File(raw_scan.path, "r") as ring_current_file:
    raw_ring_current = np.asarray(
        workflow.h5_dataset(ring_current_file, "entry/instrument/synchrotron/current")[
            ()
        ],
        dtype=float,
    ).reshape(-1)
if raw_ring_current.shape != raw_scan.angles.shape:
    raise ValueError("Ring-current readings must match the raw scan frame count")
if not np.all(np.isfinite(raw_ring_current) & (raw_ring_current > 0)):
    raise ValueError("Ring-current readings must be finite and positive")
group_ring_current = np.array(
    [raw_ring_current[members].mean() for members in frame_groups]
)
plot_group_i0 = analysis_scan.i0[contributing_indices]
plot_ring_current = group_ring_current[contributing_indices]
normalised_group_i0 = plot_group_i0 / plot_group_i0.mean()
normalised_ring_current = plot_ring_current / plot_ring_current.mean()

fig, axes = plt.subplots(2, 2, figsize=(15, 10), constrained_layout=True)
for plot_axis, field, title in (
    (axes[0, 0], "pixel_divisor", "Pixel-response divisor"),
    (axes[0, 1], "combined_divisor", "Pixel x sensor divisor"),
):
    image = plot_axis.imshow(
        calibration_fit[field],
        origin="lower",
        aspect="auto",
        cmap="coolwarm",
        vmin=0.95,
        vmax=1.05,
    )
    plot_axis.set(title=title, xlabel="Detector column", ylabel="Detector row")
    fig.colorbar(image, ax=plot_axis)
axes[1, 0].plot(np.arange(1, 9), calibration_fit["sensor_gains"], marker="o")
axes[1, 0].set(
    title="Eight sensor gains", xlabel="Sensor region", ylabel="Divisor gain"
)
gain_axis = axes[1, 1]
gain_axis.plot(
    analysis_scan.angles[contributing_indices],
    calibration_fit["frame_gains"][contributing_indices],
    marker=".",
    color="tab:blue",
    label=(
        "Refined position gain"
        if FIT_FRAME_GAINS
        else "Position gain fixed at 1 (not refined)"
    ),
)
gain_axis.axhline(1, color="black", linewidth=0.7)
gain_axis.set(
    title="Position gains and measured monitor diagnostics",
    xlabel="Angle (degrees)",
    ylabel="Gain / normalized monitor (dimensionless)",
)
gain_axis.plot(
    analysis_scan.angles[contributing_indices],
    normalised_group_i0,
    marker=".",
    color="tab:orange",
    label="Summed group I0 / mean",
)
gain_axis.plot(
    analysis_scan.angles[contributing_indices],
    normalised_ring_current,
    linestyle="--",
    marker=".",
    color="tab:green",
    label="Mean group ring current / mean",
)
gain_axis.ticklabel_format(axis="y", style="plain", useOffset=False)
gain_axis.legend(fontsize=8, loc="best")
plt.show()
````

## Q-dependent sampling on full-range processing 2θ bins

These panels use the full unmasked pyFAI pixel-centre range across all positions, with bin counts derived from `TWO_THETA_STEP_SIZE`. They are deliberately independent of `Q_MAX`, which limits only final output integrations. Edges are mapped nonlinearly to Q in **Å⁻¹**; these are not the uniform Q export grid. `FIT_Q_RANGE` and `Q_STEP_SIZE` control the separate detector-fit grid.

One observation is one retained pixel at one analysis position, assigned once by its pixel-centre 2θ after fixed masks, per-position azimuth masks and count-validity checks. Counts are not photons or fractional pyFAI integration weights. The cell streams the raw input read-only; aggregated pixels are rejected if any constituent is invalid. Raw-frame equivalents multiply observations by group size and do not imply independent angular coverage.

````python
# Full source reread: final masks and final theta bins, not the fit-only cache.
assert isinstance(THETA_BIN_COUNT, int | np.integer) and THETA_BIN_COUNT > 0
assert physical_model.azimuth_range == AZIMUTH_RANGE_DEG, (
    "Rerun from geometry setup after changing azimuth settings."
)
q_final_mask = (
    base_mask.astype(bool)
    | calibration_fit["bad_pixel_mask"].astype(bool)
    | analysis_scan.native_mask.astype(bool)
)
q_theta_range_deg = workflow.theta_range_for_scan(
    analysis_scan, physical_model, q_final_mask, theta_max=TWO_THETA_MAX_DEG
)
q_theta_edges_deg = np.linspace(
    q_theta_range_deg[0], q_theta_range_deg[1], THETA_BIN_COUNT + 1
)
q_wavelength_angstrom = float(physical_model.geometry.wavelength) * 1e10
assert np.isfinite(q_wavelength_angstrom) and q_wavelength_angstrom > 0
q_diagnostic_edges = (
    4 * np.pi * np.sin(np.deg2rad(q_theta_edges_deg) / 2) / q_wavelength_angstrom
)
q_diagnostic_indices = np.arange(len(analysis_scan.angles), dtype=int)
assert len(frame_groups) == len(q_diagnostic_indices)
q_observations_by_position = np.zeros(
    (len(q_diagnostic_indices), THETA_BIN_COUNT), dtype=np.int64
)
q_source_group_sizes = np.array(
    [len(members) for members in frame_groups], dtype=np.int64
)
q_expected_observations = 0
q_expected_raw_equivalents = 0
q_statistics_start = time.perf_counter()
with h5py.File(raw_scan.detector_path, "r") as q_source_file:
    q_source_data = workflow.h5_dataset(q_source_file, raw_scan.detector_dataset)
    for position, index in enumerate(q_diagnostic_indices):
        q_members = frame_groups[int(index)]
        if USE_ANGLE_AGGREGATION:
            q_counts = np.zeros(analysis_scan.shape, dtype=np.float64)
            q_invalid_any = np.zeros(analysis_scan.shape, dtype=bool)
            for q_frame_index in q_members:
                q_frame = np.asarray(
                    q_source_data[int(q_frame_index)], dtype=np.float64
                )
                q_invalid = (
                    ~np.isfinite(q_frame) | (q_frame < 0) | (q_frame >= RAW_SATURATION)
                )
                q_invalid_any |= q_invalid
                q_counts += np.where(q_invalid, 0.0, q_frame)
            q_counts[q_invalid_any] = np.nan
        else:
            assert len(q_members) == 1
            q_counts = np.asarray(q_source_data[int(q_members[0])], dtype=np.float64)
        # Same count-validity and per-position chi mask as integrate_frame.
        q_angle = float(analysis_scan.angles[index])
        q_integrator = physical_model.geometry.get_ai(q_angle)
        q_bad = (
            q_final_mask
            | workflow.azimuth_mask(physical_model, q_angle, q_integrator)
            | ~np.isfinite(q_counts)
            | (q_counts < 0)
            | (q_counts >= cache_saturation)
        )
        q_theta_pixels_deg = np.rad2deg(
            q_integrator.center_array(analysis_scan.shape, unit="2th_rad")
        )
        q_valid = (
            ~q_bad
            & np.isfinite(q_theta_pixels_deg)
            & (q_theta_pixels_deg >= q_theta_edges_deg[0])
            & (q_theta_pixels_deg <= q_theta_edges_deg[-1])
        )
        q_retained_theta = q_theta_pixels_deg[q_valid]
        q_observations_by_position[position] = np.histogram(
            q_retained_theta, bins=q_theta_edges_deg
        )[0]
        q_retained_number = int(np.count_nonzero(q_valid))
        q_expected_observations += q_retained_number
        q_expected_raw_equivalents += q_retained_number * len(q_members)
        assert int(q_observations_by_position[position].sum()) == q_retained_number
q_observation_counts = q_observations_by_position.sum(axis=0)
q_raw_frame_equivalents = q_source_group_sizes @ q_observations_by_position
q_position_coverage = np.count_nonzero(q_observations_by_position, axis=0)
assert int(q_observation_counts.sum()) == q_expected_observations
assert int(q_raw_frame_equivalents.sum()) == q_expected_raw_equivalents
assert np.all(q_raw_frame_equivalents >= q_observation_counts)
assert np.all(q_position_coverage <= len(q_diagnostic_indices))


def q_to_tth(q):
    # Clip for Matplotlib's limit evaluation outside the physical Q domain.
    ratio = q * q_wavelength_angstrom / (4 * np.pi)
    return 2 * np.rad2deg(np.arcsin(np.clip(ratio, -1, 1)))


def tth_to_q(tth):
    return 4 * np.pi * np.sin(np.deg2rad(tth) / 2) / q_wavelength_angstrom


fig_q_statistics = plt.figure(figsize=(15, 4.5), constrained_layout=True)
q_statistics_grid = fig_q_statistics.add_gridspec(1, 2)
q_count_axis = fig_q_statistics.add_subplot(q_statistics_grid[0, 0])
q_coverage_axis = fig_q_statistics.add_subplot(q_statistics_grid[0, 1])
q_statistics_axes = (q_count_axis, q_coverage_axis)
q_count_axis.stairs(
    q_observation_counts,
    q_diagnostic_edges,
    color="tab:blue",
    label="Retained pixel × analysis-position observations",
)
if np.any(q_source_group_sizes > 1):
    q_count_axis.stairs(
        q_raw_frame_equivalents,
        q_diagnostic_edges,
        color="tab:orange",
        label="Retained raw-frame equivalents (approx. group geometry)",
    )
    q_count_axis.set_yscale("symlog", linthresh=1)
q_count_axis.set(
    title="A · Observations before integration · final θ bins",
    ylabel="Pixel observations",
)
q_count_axis.legend(fontsize=8)
q_coverage_axis.stairs(
    q_position_coverage, q_diagnostic_edges, color="tab:green", fill=True, alpha=0.6
)
q_coverage_axis.set(
    title="B · Analysis-position coverage · final θ bins",
    ylabel="Positions contributing ≥1 pixel",
    ylim=(0, max(1, len(q_diagnostic_indices)) * 1.1),
)
q_coverage_axis.axhline(
    len(q_diagnostic_indices),
    color="0.4",
    linestyle="--",
    linewidth=0.8,
    label=f"{len(q_diagnostic_indices)} analysis positions total",
)
q_coverage_axis.legend(fontsize=8)
for q_axis in q_statistics_axes:
    q_axis.set(
        xlabel=r"Q ($\mathrm{\AA}^{-1}$)",
        xlim=(q_diagnostic_edges[0], q_diagnostic_edges[-1]),
    )
    q_axis.spines[["top", "right"]].set_visible(False)
    q_axis.secondary_xaxis("top", functions=(q_to_tth, tth_to_q)).set_xlabel(
        "2θ (degrees)"
    )
    q_axis.grid(axis="y", alpha=0.2)
fig_q_statistics.suptitle(
    f"{SCAN_PATH.stem} · Full-range pre-integration sampling after fixed and "
    "azimuth masks\n"
    f"{THETA_BIN_COUNT} final 2θ bins over {q_theta_range_deg}°; "
    f"Δ2θ = {q_theta_edges_deg[1] - q_theta_edges_deg[0]:.4g}°; "
    "nonlinear Q edges in Å⁻¹"
)
plt.show()
print("Azimuth range (pyFAI chi degrees):", physical_model.azimuth_range)
print("Final 2theta bins / limits (degrees):", THETA_BIN_COUNT, q_theta_range_deg)
print("Q limits (Å⁻¹):", q_diagnostic_edges[[0, -1]])
print(
    "Source raw frames / analysis positions:",
    len(raw_scan.angles),
    len(q_diagnostic_indices),
)
print("Total retained pixel observations:", f"{q_observation_counts.sum():,}")
print(
    "Retained raw-frame equivalents (group geometry approximation):",
    f"{q_raw_frame_equivalents.sum():,}",
)
print(
    "Bins with no observations:",
    int(np.count_nonzero(q_observation_counts == 0)),
    "/",
    THETA_BIN_COUNT,
)
print(
    "Observations per final bin (minimum / median / maximum):",
    int(q_observation_counts.min()),
    float(np.median(q_observation_counts)),
    int(q_observation_counts.max()),
)
print(
    "Contributing positions per final bin (minimum / median / maximum):",
    int(q_position_coverage.min()),
    float(np.median(q_position_coverage)),
    int(q_position_coverage.max()),
)
print(
    "Observation / raw-equivalent conservation checks passed; elapsed (s):",
    time.perf_counter() - q_statistics_start,
)
````

## Stage 2 — full-range scales, then final-only Q and 2theta extraction

With `REFINE_ALL_POSITION_SCALES=True`, freeze the stage-1 detector response and integrate all positions at unity scales over the **entire unmasked scan range**. Same-bin overlaps determine one intensity divisor per position, replacing rather than multiplying stage-1 nuisance gains. `Q_MAX` never enters this scaling integration.

Inverse-log-variance-weighted graph least squares requires a connected graph with `OVERLAP_MIN_PIXEL_WEIGHT` and `OVERLAP_MIN_BINS` support, with equal-position mean log gain zero. Restrictive azimuth selection can still disconnect the graph; no thresholds are silently relaxed. Signal, raw-count variance and pixel counts remain invariant: only normalization accumulators are multiplied by the fitted gains. Overlap improvement is in-sample, and Poisson sigma excludes scale/model uncertainty and correlated-bin effects.

After scaling, independently integrate final uniform Q (Å⁻¹) and 2theta (degrees) grids with target widths `Q_STEP_SIZE` and `TWO_THETA_STEP_SIZE`. Only these integrations use `Q_MAX`, resolved from the explicit parameter, NeXus metadata, or unmasked pyFAI maximum. Final gains and detector maps remain frozen. Both grids are saved in the pattern HDF5 alongside the calibration when `WRITE_CALIBRATION=True`; `WRITE_XY=True` saves separate `_q.xy` and `_tth.xy` files. Existing outputs are never overwritten.

Overlap comparisons and raw audits below remain **full range**, so lowering the final output limit does not remove their support. The direct final-intensity plot uses the cropped exported data. No smoothing, tapering, additive offsets or Q-dependent scale is introduced. Detector-response use outside the fit Q range remains extrapolation.

````python
# Stage 2 uses the FULL scan; final output is a separate, cropped integration.
calibration_file_written = None
assert calibration_fit["fit_frame_gains"] == FIT_FRAME_GAINS
assert fit_settings["fit_frame_gains"] == FIT_FRAME_GAINS
for output_path, enabled in (
    (CALIBRATION_OUTPUT, WRITE_CALIBRATION),
    (PATTERN_OUTPUT, WRITE_CALIBRATION),
    (TWO_THETA_XY_OUTPUT, WRITE_XY),
    (Q_XY_OUTPUT, WRITE_XY),
):
    if enabled:
        assert not output_path.exists(), f"Choose a new output name: {output_path}"
try:
    detector_calibration = {
        "fit_settings": fit_settings,
        "source_scan": str(analysis_scan.path),
        "frame_angles": analysis_scan.angles,
        "frame_i0": analysis_scan.i0,
        "frame_gains": calibration_fit["frame_gains"],
        "combined_divisor": calibration_fit["combined_divisor"],
        "base_mask": base_mask,
        "bad_pixel_mask": calibration_fit["bad_pixel_mask"],
    }
    processed_without_scales = workflow.process_scan(
        analysis_scan,
        detector_calibration,
        physical_model,
        q_range=FIT_Q_RANGE_NM,
        q_npt=FIT_BIN_COUNT,
        theta_npt=THETA_BIN_COUNT,
        theta_range=processing_theta_range,
        saturation=cache_saturation,
    )
    processed_before_scales = workflow.apply_position_scales(
        processed_without_scales, calibration_fit["frame_gains"]
    )
    position_scale_fit = None
    if REFINE_ALL_POSITION_SCALES:
        position_scale_fit = workflow.refine_position_scales(
            processed_without_scales["groups"]["merged_2theta"],
            minimum_pixel_weight=OVERLAP_MIN_PIXEL_WEIGHT,
            minimum_overlap_bins=OVERLAP_MIN_BINS,
            fractional_floor=OVERLAP_FRACTIONAL_FLOOR,
        )
        final_position_gains = position_scale_fit["frame_gains"]
        print(
            "Connected graph positions / edges:",
            position_scale_fit["n_positions"],
            position_scale_fit["n_edges"],
        )
        print(
            "Weighted pair RMS before / after:",
            position_scale_fit["weighted_edge_rms_before"],
            position_scale_fit["weighted_edge_rms_after"],
        )
    else:
        final_position_gains = calibration_fit["frame_gains"].copy()
    # Full-range results remain available for overlap diagnostics and raw audits.
    processed_for_xy = workflow.apply_position_scales(
        processed_without_scales, final_position_gains
    )
    processed_for_xy["position_scale_model"] = MODEL_LABEL
    print(
        "Final position divisor min / max:",
        final_position_gains.min(),
        final_position_gains.max(),
    )
    print("position motor_deg stage1_in_fit stage1_gain final_gain")
    for scale_i, scale_angle in enumerate(analysis_scan.angles):
        print(
            scale_i,
            scale_angle,
            scale_i in calibration_fit["fit_indices"],
            calibration_fit["frame_gains"][scale_i],
            final_position_gains[scale_i],
        )
    final_fit_settings = {
        **fit_settings,
        "q_range_unit": "nm^-1",
        "q_step_size_angstrom_inverse": Q_STEP_SIZE,
        "two_theta_step_size_deg": TWO_THETA_STEP_SIZE,
        "position_scale_model": MODEL_LABEL,
        "refine_all_position_scales": bool(REFINE_ALL_POSITION_SCALES),
    }
    if position_scale_fit is not None:
        final_fit_settings["position_scale_refinement"] = {
            **position_scale_fit["settings"],
            "range": ("full unmasked scan 2theta range; independent of output crop"),
            "theta_range_deg": list(processing_theta_range),
            "gauge": position_scale_fit["gauge"],
            "replaces_stage1_gains": True,
        }
    export_fit = {
        **calibration_fit,
        "frame_gains": final_position_gains.copy(),
        "q_range": FIT_Q_RANGE_NM,
    }
    final_output = integrate_output(
        analysis_scan,
        physical_model,
        export_fit,
        base_mask,
        final_theta_range,
        TWO_THETA_STEP_SIZE,
        Q_STEP_SIZE,
        final_position_gains,
        saturation=cache_saturation,
    )
    if WRITE_CALIBRATION:
        calibration_file_written = workflow.save_calibration(
            CALIBRATION_OUTPUT,
            analysis_scan,
            physical_model,
            export_fit,
            base_mask,
            sensor_ids,
            final_fit_settings,
            validation_reports,
        )
        with h5py.File(calibration_file_written, "r+") as stage_file:
            stage_diag = workflow.h5_group(stage_file, "fit_diagnostics")
            stage_diag.create_dataset(
                "detector_fit_nuisance_frame_gains",
                data=calibration_fit["frame_gains"],
            )
            stage_diag.attrs["frame_gains_stage"] = (
                "full-range overlaps"
                if REFINE_ALL_POSITION_SCALES
                else "detector-fit range"
            )
            stage_diag.attrs["validation_scope"] = (
                "stage 1 detector fit only; stage 2 residuals are in-sample"
            )
            if position_scale_fit is not None:
                stage_report = {
                    k: v for k, v in position_scale_fit.items() if k != "frame_gains"
                }
                stage_diag.attrs["position_scale_refinement_json"] = json.dumps(
                    stage_report
                )
                stage_diag.attrs["scale_uncertainty"] = (
                    "not propagated; correlated overlap edges and split bins"
                )
        saved = workflow.load_calibration(calibration_file_written)
        np.testing.assert_array_equal(saved["frame_gains"], final_position_gains)
        np.testing.assert_array_equal(
            saved["combined_divisor"], calibration_fit["combined_divisor"]
        )
        workflow.save_processed(
            PATTERN_OUTPUT, analysis_scan, saved, physical_model, final_output
        )
        print(
            "Created calibration / dual-grid pattern:",
            CALIBRATION_OUTPUT,
            PATTERN_OUTPUT,
        )
    for output_group_name, output_path, output_label in (
        ("merged_2theta", TWO_THETA_XY_OUTPUT, "2theta_deg"),
        ("merged_q", Q_XY_OUTPUT, "q_A^-1"),
    ):
        output_group = final_output["groups"][output_group_name]
        supported = (output_group["accumulators"][1] > 0) & np.isfinite(
            output_group["intensity"]
        )
        output_xy = np.column_stack(
            (
                output_group["axis"][supported],
                output_group["intensity"][supported],
            )
        )
        assert output_xy.size > 0, f"No supported bins in {output_group_name}"
        if output_group_name == "merged_2theta":
            xy_data = output_xy
        if WRITE_XY:
            with output_path.open("x", encoding="ascii") as xy_file:
                np.savetxt(
                    xy_file,
                    output_xy,
                    fmt="%.10g",
                    header=(
                        f"{output_label} intensity; model={MODEL_LABEL}; "
                        "measured I0 retained"
                    ),
                )
            np.testing.assert_allclose(
                np.loadtxt(output_path, ndmin=2),
                output_xy,
                rtol=6e-10,
                atol=1e-12,
            )
            print("Created XY:", output_path)
finally:
    if working_directory is not None:
        working_directory.cleanup()
        print("Removed temporary aggregated detector data.")
````
